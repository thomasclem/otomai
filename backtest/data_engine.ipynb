# Download Historical data into CSV in Database

```python
exchange_ = ccxt.binance()
```

```python
market = exchange_.load_markets()
```

```python
market_filter = {k:v for k,v in market.items()if k.endswith(":USDT")}
```

```python
market_filter
```

```python
from datetime import datetime

input_date = "2024-01-01 00:00:00"
input_date_ms = int(datetime.strptime(input_date, "%Y-%m-%d %H:%M:%S").timestamp() * 1000)

symbols = [
    s for s in market.keys()
    if s.endswith(":USDT")
    and not s.startswith("NC")
    and market[s].get("created") is not None
    and int(market[s].get("created")) > input_date_ms
]
```

```python
len(symbols)
```

```python
exchange_name = "binance"

intervals = ["1h"]

coin_to_dl = [
    "BTC/USDT:USDT",
]

exchange = ExchangeDataManager(
    exchange_name=exchange_name, path_download="database/exchanges"
)

await exchange.download_data(
    coins=symbols,
    intervals=intervals,
    start_date=input_date
)
```

```python
df = exchange.load_data(
    coin="ETH/USDT:USDT",
    interval="1h"
)

df
```
